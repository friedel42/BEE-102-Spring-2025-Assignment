**Building Markov transition matrix**

You have learned the key differences between a simple dinucleotide frequency model and a Markov model for modeling DNA sequences.
The task is to construct a Markov Model transition matrix, where the probability of each nucleotide following another is calculated and stored in a dictionary (e.g. P ( C | A ) , P ( G | T ) , etc.).

Input:

A string containing { ' A ' , ' G ' , ' C ' , ' T ' } (the DNA sequence).

Output:

A matrix containing the transition probabilites from one nucleotide to another.

```python
from collections import defaultdict

sequence = "ACGTTGCAAGTTCG"
nucleotides = ['A', 'C', 'G', 'T']
counts = {x: defaultdict(int) for x in nucleotides}
probs = {x: {} for x in nucleotides}

for i in range(len(sequence) - 1):
    curr = sequence[i]
    next = sequence[i + 1]
    if curr in nucleotides and next in nucleotides:
        counts[curr][next] += 1

for curr in nucleotides:
    total = sum(counts[curr].values())
    for next in nucleotides:
        probs[curr][next] = counts[curr][next] / total if total > 0 else 0.0

print(f"{'':^5}", end='')
for n in nucleotides:
    print(f"{n:^12}", end='')
print('\n')

for curr in nucleotides:
    print(f"{curr:^4}", end='')
    for next in nucleotides:
        print(f"{probs[curr][next]:^12.4f}", end='')
    print()
```
